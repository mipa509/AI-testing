# Benchmark Setup Guide

Use this notebook as the runbook for consistent model benchmarking.

- Qwen3.5 27B

## Execution Protocol

**Runtime condition (important):** All benchmark tests were run using **Ollama in terminal** sessions.\nReplications should use the same approach to keep test conditions consistent across models.\n
For every task, run each model under the same conditions:

1. Use the exact same task prompt.
2. Use a fresh session per model.
3. Do not provide corrective guidance on first pass.
4. Save raw outputs to `Evaluation N.md`.
5. Run a second-pass correction prompt only after first-pass outputs are captured for all models.

## Scoring Protocol

Score each model output from 1 to 5 on:

- Correctness
- Robustness to messy real-world engineering data
- Readability
- Maintainability / modularity
- Engineering suitability and judgement
- Clarity of presented information

Always include:

- criterion-level winner
- overall task winner
- key weaknesses per model
- practical significance (minor / moderate / large).

## Evaluator Prompt (Template)

```text
You are acting as a strict coding evaluator for structural engineering automation tasks.

Rules:
1) Do not rewrite solutions unless asked.
2) Evaluate correctness, robustness, readability, maintainability, engineering suitability, and clarity.
3) Be skeptical of unit errors, wrong section properties, fragile data handling, and invented logic.
4) Prefer production-safe, auditable engineering code.
5) Score each criterion from 1 to 5 for each model.
6) Provide criterion winners, overall winner, and key weaknesses.
7) State practical significance of the difference.
```
